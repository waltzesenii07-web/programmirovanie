```python
import pygame
import math
import random

pygame.init()

width, height = 400, 300
screen = pygame.display.set_mode((width, height), pygame.RESIZABLE)
pygame.display.set_caption("игра смайлик")

pygame.mixer.init()
pygame.mixer.music.load("fon_music.mp3")
pygame.mixer.music.play(-1)
score_sound = pygame.mixer.Sound("winner.mp3")

x, y = width // 4, (height - 3 * height // 4)
gravity = 2
y_speed = 0
x_speed = 0
score = 0

font = pygame.font.Font(None, 36)

background_original = pygame.image.load("fon.png")
smile_image = pygame.image.load("smile.png").convert()
smile_image.set_colorkey((255, 255, 255))
background_original = pygame.transform.scale(background_original, (width, height))
smile_image = pygame.transform.scale(smile_image, (width//5, height//5))
background = background_original

enemy_images = [
    pygame.image.load("1.png").convert_alpha(),  
    pygame.image.load("2.png").convert_alpha(),
    pygame.image.load("3.png").convert_alpha(),
    pygame.image.load("4.png").convert_alpha(),
    pygame.image.load("5.png").convert_alpha()
]

for i in range(len(enemy_images)):
    enemy_images[i] = pygame.transform.scale(enemy_images[i], (width//7, height//7))

objects = []
spawn_timer = 0
spawn_interval = 100
object_speed = -1

running = True
clock = pygame.time.Clock()

scored_flag = False
score = 0


while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.VIDEORESIZE:
            width, height = event.w, event.h
            screen = pygame.display.set_mode((width, height), pygame.RESIZABLE)
            background = pygame.transform.scale(background_original, (width, height))
        elif event.type == pygame.KEYDOWN:
            if event.key == pygame.K_w:
                y_speed = -20
    
            elif event.key == pygame.K_d:
                x_speed = 5
                
        elif event.type == pygame.KEYUP:
            if event.key in (pygame.K_a, pygame.K_d):
                x_speed = -5
                
        elif event.type == pygame.KEYDOWN:
            if event.key == pygame.K_s:
                y_speed = -20
       
            elif event.key == pygame.K_z:
                x_speed = -5
            elif event.key == pygame.K_c:
                x_speed = 5

        elif event.type == pygame.KEYUP:
            if event.key in (pygame.K_z, pygame.K_c):
                x_speed = 0

    spawn_timer += 1
    if spawn_timer >= spawn_interval:
        enemy_index = random.randint(0, len(enemy_images) - 1) 
        spawn_y = height - 100
        objects.append([width, spawn_y, enemy_index]) 
        spawn_timer = 0

    y_speed += gravity
    y += y_speed

    smile_rect = pygame.Rect(x, y, width//5, height//5)
    enemy_rect = pygame.Rect(x, spawn_y, width//7, height//7)

    if enemy_rect.colliderect(smile_rect) and not scored_flag:
        scored_flag = True
        score += 1
        score_sound.play()

    if y + smile_image.get_height() // 2 >= height:
        y = height - smile_image.get_height() // 2
        y_speed = 0
    if y - smile_image.get_height() // 2 <= 0:
        y = smile_image.get_height() // 2
        y_speed = 0

    x += x_speed
  
    for obj in objects[:]:
        obj[0] += object_speed
        obj[1] += random.uniform(-1, 1)
        if obj[0] < -50:
            objects.remove(obj)

    bg_x = -x % width
    screen.blit(background, (bg_x, 0))
    screen.blit(background, (bg_x - width, 0))
    screen.blit(smile_image, (width//4 - smile_image.get_width()//2, int(y) - smile_image.get_height()//2))
    
    for obj in objects:
        img = enemy_images[obj[2]]
        screen.blit(img, (obj[0], obj[1]))
    
    score_text = font.render(f"Score: {score}", True, (255, 255, 255))
    screen.blit(score_text, (10, 10))
    
    pygame.display.flip()
    pygame.time.delay(10)

pygame.quit()
```
